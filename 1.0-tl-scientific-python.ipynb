## Task 4

Convert the RGB image into the HSV color space. Rotate the colors in the HSV color space, convert it back to the RGB color space and plot it together with its histogram. For this task, you are only allowed to use NumPy.

More information about the HSV color space: https://en.wikipedia.org/wiki/HSL_and_HSV

````python
def rgb2hsv(image):
  """Convert the rgb image into hsv color space and return it.

  image -- rgb image
  """
  # TODO: implement function
  img = image.astype(np.float64)
  if img.max() > 1.0:
    img = img / 255.0

  r = img[:, :, 0]
  g = img[:, :, 1]
  b = img[:, :, 2]

  maxc = np.maximum(np.maximum(r, g), b)
  minc = np.minimum(np.minimum(r, g), b)
  v = maxc

  deltac = maxc - minc
  s = np.zeros_like(v)
  s[maxc > 0] = deltac[maxc > 0] / maxc[maxc > 0]

  h = np.zeros_like(v)
  nz = deltac != 0

  mask_r = (r == maxc) & nz
  mask_g = (g == maxc) & nz & (~mask_r)
  mask_b = (b == maxc) & nz & (~mask_r) & (~mask_g)

  h[mask_r] = (g[mask_r] - b[mask_r]) / deltac[mask_r]
  h[mask_g] = 2.0 + (b[mask_g] - r[mask_g]) / deltac[mask_g]
  h[mask_b] = 4.0 + (r[mask_b] - g[mask_b]) / deltac[mask_b]

  h = (h / 6.0) % 1.0

  hsv = np.stack([h, s, v], axis=-1)
  return hsv

def hsv2rgb(image):
    '''
    Convert the hsv image into rgb color space and return it.

    image -- rgb image
    '''
    # TODO: implement function
    h = image[:, :, 0]
    s = image[:, :, 1]
    v = image[:, :, 2]

    i = (h * 6.0).astype(int) % 6
    f = (h * 6.0) - i

    p = v * (1.0 - s)
    q = v * (1.0 - s * f)
    t = v * (1.0 - s * (1.0 - f))

    r = np.zeros_like(h)
    g = np.zeros_like(h)
    b = np.zeros_like(h)

    m0 = i == 0
    m1 = i == 1
    m2 = i == 2
    m3 = i == 3
    m4 = i == 4
    m5 = i == 5

    r[m0], g[m0], b[m0] = v[m0], t[m0], p[m0]
    r[m1], g[m1], b[m1] = q[m1], v[m1], p[m1]
    r[m2], g[m2], b[m2] = p[m2], v[m2], t[m2]
    r[m3], g[m3], b[m3] = p[m3], q[m3], v[m3]
    r[m4], g[m4], b[m4] = t[m4], p[m4], v[m4]
    r[m5], g[m5], b[m5] = v[m5], p[m5], q[m5]

    rgb = np.clip(np.stack([r, g, b], axis=-1) * 255.0, 0, 255)
    return rgb.astype(np.uint8)
````

````python
# Change the color space from rgb to hsv and plot it
image_hsv = rgb2hsv(image)

# TODO: plot image
plt.figure(figsize=(6, 5))
plt.imshow(image_hsv)
plt.title('HSV Image')
plt.axis('off')
plt.show()

# TODO: Rotate the colors in hsv color space
image_hsv_rotated = image_hsv.copy()
image_hsv_rotated[:, :, 0] = (image_hsv_rotated[:, :, 0] + 0.33) % 1.0

# TODO: Now, convert it back from HSV to RGB and plot it
image_rgb_rotated = hsv2rgb(image_hsv_rotated)

plt.figure(figsize=(6, 5))
plt.imshow(image_rgb_rotated)
plt.title('Rotated RGB Image')
plt.axis('off')
plt.show()

# TODO: Plot the color histogram of the inverted rgb image
plot_color_histogram(image_rgb_rotated)
````
